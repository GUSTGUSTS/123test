```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/results/exp2/weights/best.pt")

results = model.predict(
    source="/content/drive/MyDrive/test/pics",
    save=True
)
```

```python
print("--- สรุปผลการทำนาย ---")
for i, r in enumerate(results):
    # ดึง index ของ 2 อันดับแรก
    top1_index = r.probs.top5[0]
    top2_index = r.probs.top5[1]

    # ดึงชื่อคลาส (เลขที่ทำนายได้)
    top1_name = r.names[top1_index]
    top2_name = r.names[top2_index]

    # ดึงค่าความมั่นใจ
    top1_conf = r.probs.top5conf[0].item()
    top2_conf = r.probs.top5conf[1].item()

    print(f"ภาพที่ {i+1}: น่าจะอยู่ที่ '{top1_name}' (ความมั่นใจ {top1_conf:.2f})")

    # ตรวจสอบส่วนต่างความน่าจะเป็น
    if (top1_conf - top2_conf) < 0.20:
        print(f"**ภาพนี้อาจจะอยู่ระหว่าง '{top1_name}' กับ '{top2_name}'")
```

```python
with open('/content/drive/MyDrive/test/check.txt', 'r', encoding='utf-8') as f:
    ground_truths = [line.strip() for line in f.readlines()]

print("--- เปรียบเทียบผลการทำนายกับเฉลย ---")
correct_count = 0

# ใช้ zip เพื่อลูป results และ ground_truths ไปพร้อมๆ กัน
for i, (r, truth) in enumerate(zip(results, ground_truths)):
    # ดึงชื่อคลาส (เลขที่ทำนายได้อันดับ 1)
    top1_index = r.probs.top5[0]
    top1_name = r.names[top1_index]

    # ตรวจสอบว่าถูกหรือผิด
    if top1_name == truth:
        status = "ถูก"
        correct_count += 1
    else:
        status = "ผิด"

    print(f"ภาพที่ {i+1} -> ทาย: {top1_name} : เฉลย: {truth} : สรุป: {status}")

print("-" * 40)
print(f"สรุปความถูกต้องทั้งหมด: {correct_count}/{len(results)} ภาพ คิดเป็น {correct_count/len(results)*100:.2f}%")
```
